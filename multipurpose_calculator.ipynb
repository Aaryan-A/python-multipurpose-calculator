```python
# Welcome the user and tell the code's functionality
print("Hello User")
print("This is the multipurpose calculator")
print()

# Display options using a list
options = [
    "1. Height: Feet and inches to cm",
    "2. Weight: Lbs to Kg",
    "3. Simple Interest",
    "4. Hourly Wage to Annual Salary"
]

# This prints the code on separate lines for clarity
for option in options:
    print(option)

# Adds another line for more clarity
print()

# When user1 gets called, then the program finds that the user has chosen Height: Feet and inches to cm and converts the feet and inches to centimeters based on the users input
def user1(feet, inches):
    print()
    final1 = (feet * 12 + inches) * 2.54
    print()
    # Outputs the height in centimeters. The value that is printed is through a string
    print("Your height in centimeters is " + str(final1))

# When user2 gets called, then the program finds that the user has chosen Weight: Lbs to Kg and converts the weight in pounds to kg based on the user's input
def user2(weight):
    print()
    final2 = weight * 0.45359237
    print()
    # Outputs the weight in kgs. The value that is printed is through a string
    print("Your weight in kgs is " + str(final2))

# When user3 gets called, then the program finds that the user has chosen Simple Interest and based off of the users input, it calculates the interest amount
def user3(principle, rate_of_interest, years):
    print()
    calc2 = (principle * rate_of_interest * years) / 100
    print()
    # Outputs the total interest amount. The value that is printed is through a string
    print("Your total interest amount is " + str(calc2))

# When user4 gets called, then the program finds that the user has chosen Hourly Wage to Annual Salary and based off of the users input, it calculates the annual salary
def user4(hourly, weeks):
    print()
    calc_final = hourly * weeks * 52
    print()
    # Outputs the annual salary amount. The value that is printed is through a string
    print("Your Annual Salary is " + str(calc_final))

# Loop to allow multiple calculations
while True:
    # This asks the user which calculation they would like to do
    user_input = input("Which calculation would you like to do? (Enter the number or type 'exit' to quit) ")

    # When the user inputs 1, then the questions are asked to define the parameters and then user1 is called from before
    if user_input == "1":
        print("You have chosen, Height: Feet and inches to cm")
        # The 2 lines below get the values from the user which define the parameters for user1
        feet = int(input("What is your height in feet? (Ex: My height is 5'10, so enter 5) "))
        inches = int(input("What is your height in inches? (Ex: My height is 5'10, so enter 10) "))
        user1(feet, inches)

    # When the user inputs 2, then the question is asked to define the parameters and then user2 is called from before
    elif user_input == "2":
        print("You have chosen, Weight: Lbs to Kg")
        # The line below get the values from the user which define the parameter for user2
        weight = int(input("What is your weight in lbs? "))
        user2(weight)

    # When the user inputs 3, then the questions are asked to define the parameters and then user3 is called from before
    elif user_input == "3":
        print("You have chosen, Simple Interest")
        # The 3 lines below get the values from the user which define the parameters for user3
        principle = float(input("Enter your principal amount: "))
        rate_of_interest = float(input("Enter your rate of interest: "))
        years = float(input("Enter the time in years: "))
        user3(principle, rate_of_interest, years)

    # When the user inputs 4, then the questions are asked to define the parameters and then user4 is called from before
    elif user_input == "4":
        print("You have chosen, Hourly Wage to Annual Salary")
        # The 2 lines below get the values from the user which define the parameters for user4
        hourly = float(input("What is your hourly wage? "))
        weeks = int(input("How many weeks have you worked in a year? "))
        user4(hourly, weeks)

    # When the user inputs exit, then the program prints out goodbye and breaks the loop
    elif user_input == "exit":
        print("Goodbye!")
        break

    # When the user inputs anything else not listed, then the program prints Invalid choice and tells the user to try again
    else:
        print("Invalid choice, please try again.")
```
